Для создания ветвления в коде программы используется условный оператор:

**Пример 1**. Определить, является ли введенное пользователем число четным.

````python
num = int(input('Введите целое число: '))

# сначала пишем if, потом логическое условие, потом с отступом тело условия
# если условие истинно, будет выполняться это тело условия
# если ложно, выполнится то, что находится в ветке else

if num % 2 == 0:
    print('Число чётное')
else:
    print('Число нечётное')
````

````python
age = int(input('Введите ваш возраст: '))
threshold = 18

if age >= threshold:
  print('Добро пожаловать!')
else:
  print('Доступ запрещен!')
````

````python
x = float(input('x = '))
y = float(input('y = '))
R = float(input('R = '))

if (x * x + y * y)**0.5 <= R:
    print("Попадание")
else:
    print("Промах")
````

Решим эту задачу используя **тернарный оператор** Python - это сокращённый формат записи условий, содержащих не более одного действия в теле `if` и `else`:

```python
значение_если_истина if условие else значение_если_ложь`
```
Пример:
```python
result = a if condition else b
```
эквивалентно:
```python
if condition:
    result = a
else
    result = b
```

````python
x = float(input('x = '))
y = float(input('y = '))
R = float(input('R = '))

if (x * x + y * y)**0.5 <= R:
    print("Попадание")
else:
    print("Промах")


print(
    'Попадание' if (x * x + y * y) **0.5 <=R else 'Промах'
)
````

### `if... elif... else...`

Расширенная форма может содержать несколько блоков `elif` с проверкой дополнительных условий.

**Пример 5**. Написать скрипт, который определяет знак целого числа и равенство нулю.

````python
num = int(input('Введите целое число: '))

if num > 0:
    print('Число положительное')
elif num < 0:
    print('Число отрицательное')
else:
    print('Число равно нулю')
````

**Пример 6**. Напишите программу, которая будет предлагать клиенту пирожок. Попробуйте предусмотреть все возможные сценарии введённых входных данных.
    

````python
ans = input('Хотите ли пирожок?')#.lower()

if ans == 'да' or ans == 'Да' or ans == 'ДА':
    print('Держите')
elif ans in ['нет', 'не хочу', 'no', 'Нет', 'НЕТ']:
    print('Тогда до скорого')
else:
    print('Я вас не понял')
````
