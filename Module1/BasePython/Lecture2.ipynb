### `if ... else`
Для создания ветвления в коде программы используется условный оператор:

````python
condition = True

if condition:
    print('Истина')
else:
    print('Ложь')
````

**Пример 1**. Определить, является ли введенное пользователем число четным.

````python
num = int(input('Введите целое число: '))

# сначала пишем if, потом логическое условие, потом с отступом тело условия
# если условие истинно, будет выполняться это тело условия
# если ложно, выполнится то, что находится в ветке else

if num % 2 == 0:
    print('Число чётное')
else:
    print('Число нечётное')
````

**Пример 2**. Написать скрипт, который запрашивает у пользователя возраст, сравнивает с порогом в переменной `threshold`. Программа должна приветствовать пользователя если его возраст не меньше порога и выводить сообщение "Доступ запрещен!" в противном случае.

````python
age = int(input('Введите ваш возраст: '))
threshold = 18

if age >= threshold:
  print('Добро пожаловать!')
else:
  print('Доступ запрещен!')
````

**Пример 3**. Выяснить, попала ли точка $(x, y)$ в круг с центром в начале координат и радиусом $R$. См. уравнение окружности.

````python
x = float(input('x = '))
y = float(input('y = '))
R = float(input('R = '))

if (x * x + y * y)**0.5 <= R:
    print("Попадание")
else:
    print("Промах")
````

Решим эту задачу используя **тернарный оператор** Python - это сокращённый формат записи условий, содержащих не более одного действия в теле `if` и `else`:

```python
значение_если_истина if условие else значение_если_ложь
```
Пример:
```python
result = a if condition else b
```
эквивалентно:
```python
if condition:
    result = a
else:
    result = b
```

````python
x = float(input('x = '))
y = float(input('y = '))
R = float(input('R = '))

if (x * x + y * y)**0.5 <= R:
    print("Попадание")
else:
    print("Промах")


print(
    'Попадание' if (x * x + y * y) **0.5 <=R else 'Промах'
)
````

**Пример 3**. Вывести последовательность натуральных нечетных чисел не превосходящих 10.

````python
def calculate_price(
    price: int, discount: int, is_vip: bool = False
) -> float:
    """
    Функция для вычисления итоговой стоимости товара с учетом скидки

    :param price: Исходная цена товара.
    :param discount_percent: Процент скидки.
    :param is_vip: Флаг, указывающий, является ли покупатель VIP-клиентом.
    Если покупатель VIP-клиент, то к скидке прибавляется 5%

    :return: Итоговая цена.
    """
    if is_vip:
      discount_total = discount + 5
    else:
      discount_total = discount

    return price * (100 - discount_total) / 100
````

### Область видимости.

Область видимости (scope) определяет доступность переменных, функций и других объектов в разных частях программы. Он определяет, где и какие имена (идентификаторы) могут быть использованы.

Поскольку имена не объявляются заранее, интерпретатор Python по местоположению операции присваивания связывает имя с конкретным пространством имен. **Место, где выполняется присваивание, определяет пространство имен, в котором будет находиться имя, а следовательно, и область его видимости.**

Правило **LEGB** описывает последовательность, в которой Python ищет имена переменных

* **L - Local** *(локальная область видимости):* это область видимости переменных внутри функции или блока кода. Переменные, определенные внутри функции или блока кода, имеют локальную область видимости и доступны только внутри этой функции или блока.
* **E - Enclosing** *(внешняя область видимости):* если функция определена внутри другой функции, то внутренняя функция имеет доступ к переменным внешней функции.
* **G - Global** *(глобальная область видимости):* это область видимости переменных, определенных на верхнем уровне модуля. Переменные, функции и другие объекты, определенные на этом уровне, могут быть доступны из любой части программы.
* **B - Built-in** *(встроенная область видимости):* это область видимости встроенных имен, которые являются частью стандартной библиотеки Python. Эти имена доступны в любой части программы без необходимости импортирования.

````python
# Global scope
x = 0

def outer():
    # enclosed scope
    x = 1 # local scope функции outer
    def inner(): # local scope функции outer
        # local scope
        x = 2
        print(f'inner func x = {x}')
    inner() # print(2)
    print(f'outer func x = {x}') # print(1)
outer()
# встроенная функция len существует в Built-in scope
# print(len([1, 2, 3]))  # built-in
````
