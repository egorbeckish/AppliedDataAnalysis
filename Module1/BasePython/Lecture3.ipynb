**Внимание!** Рассмотренные методы не меняют исходное множество. Следующие методы изменяют множество:
```
 'add',
 'clear',
 'difference_update',
 'discard',
 'intersection_update',
 'pop',
 'remove',
 'symmetric_difference_update',
 'update'
 ```
 По понятным причинам данные методы отсутствуют в frozenset

* set.add(elem) - добавляет элемент в множество.
* set.discard(elem) - удаляет элемент, если он находится в множестве.
* set.remove(elem) - удаляет элемент из множества. KeyError, если такого элемента не существует.
* set.pop() - удаляет первый элемент из множества. Так как множества не упорядочены, нельзя точно сказать, какой элемент будет первым.
* set.clear() - очистка множества.

Словарь (ассоциативный массив, отображение) - отображение хешируемых значений (неизменяемые объекты) на произвольные объекты. Словари являются изменяемыми объектами.

Словарь, другими словами, это последовательность пар ключей и значений.

````python
a = dict(one=1, two=2, three=3)
b = {'one': 1, 'two': 2, 'three': 3}
c = dict(zip(['one', 'two', 'three'], [1, 2, 3, 4]))
d = dict([('two', 2), ('one', 1), ('three', 3)])
e = dict({'three': 3, 'one': 1, 'two': 2})
a == b == c == d == e
c
````

````python
my_dict = {
    1: 23,
    '1': '23',
    (1,): 9,
    sum: [1, 2, 4]
    # [1, 2]: 5
}
print(
    my_dict[1], my_dict['1'], my_dict[(1,)], my_dict[sum]
)
````

````python
list(zip(['one', 'two', 'three'], [1, 2, 3, 4]))
````

### Операции со словарем

|Операция/Функция|Описание|
| :---| :---|
|`len(d)`|Количество элементов в словаре.|
|`d[key]`|Возвращает элемент по ключу `key`, <br>ошибку KeyError, если такого ключа не существует.|
|`d[key] = value`|Изменяет или добавляет значение с ключом `key`.|
|`del d[key]`|Удаление ключа или ошибка.
|`key in d`|Возвращает `True`, если ключ есть в словаре.|
|`key not in d`|Возвращает `True`, если ключа нет в словаре.|
|`iter(d)`|Возвращает итератор по ключам.|

### Циклы и словари

````python
week = {
    0:'Понедельник',
    1:'Вторник',
    2:'Среда',
    3:'Четверг',
    4:'Пятница',
    5:'Суббота',
    6:'Воскресенье'
}
````

````python
list(iter(week)) # как с range
````

````python
for i in iter(week):
    print(i, week[i])
````

````python
for i in week:
    print(i, week[i])
````

````python
# частный случай
for i in range(len(week)):
    print(i, week[i])
````

### Сортировка словаря

````python
week_1 = {
    2:'Среда',
    3:'Четверг',
    5:'Суббота',
    1:'Вторник',
    6:'Воскресенье',
    0:'Понедельник',
    4:'Пятница',
}
````

````python
# По ключам
print(week_1.keys())
print(sorted(week_1.keys()))

for i in sorted(week_1.keys()):
    print(i, week_1[i])
````

````python
week_1.get
````

````python
# По значениям
print(sorted(week_1, key=week_1.get))

for i in sorted(week_1, key=week_1.get):
    print(i, week_1[i])
````

````python
sales1 = {"apple": 10, "banana": 5, "orange": 7}
sales2 = {"banana": 3, "orange": 2, "kiwi": 4}

###

total_sales = sales1.copy()
for k, v in sales2.items():
  total_sales[k] = total_sales.get(k, 0) + v

total_sales
````

### Упражнение 7.

Есть список координат точек (x, y). Выведите только те точки, которые находятся внутри квадрата с вершинами (0, 0) и (5, 5).
Координаты нельзя изменять.

````python
points = [(1, 2), (4, 5), (5, 1), (6, 3), (0, 0)]

###
for point in points:
  x, y = point
  if 0 <= x <= 5 and 0 <= y <= 5:
    print(point, 'внутри!')
  else:
    print(point, 'снаружи!')
````

### Упражнение 9.

Дан список слов. Подсчитайте, сколько раз каждое слово встречается.

````python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

###
freq_dict = {}
vocab = list(set(words))
for word in vocab:
  freq_dict[word] = words.count(word)

freq_dict
````
