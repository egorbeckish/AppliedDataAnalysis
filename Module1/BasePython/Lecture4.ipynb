## 🟢 1. Базовые понятия

**Объектно-ориентированное программирование (ООП)** используется для того, чтобы *структурировать код, делая его более удобным для понимания, сопровождения и расширения*.

> Основная идея заключается в том, чтобы объединять данные и поведение в единые сущности — **объекты**. Такой подход позволяет моделировать реальные процессы и объекты, а также повторно использовать код.  

> Объекты объединяют в себе **свойства** (данные) и **методы** (функции) их обработки. Разные объекты могут иметь разные состояния, но как правило, имеют одни и те же методы. Класс можно представлять себе как общий шаблон для создания объектов. В python класс также является объектом!

**Наглядные примеры:**

1. Конкретная автомашина, например, ваша - является объектом
реального мира, все автомашины представляют собой один класс.
2. Точка на плоскости – это класс, конкретная точка, например, М(1, - 2)
– представитель класса (объект).
3. Страница веб-сайта – класс, главная страница вашего сайта – объект.
4. В реальном мире классы структурированы иерархически, например,
автомашина может быть потомком более широкого класса –
механизм.


**Ключевые понятия:**

- **Класс** — это шаблон или описание, по которому создаются объекты. Класс определяет, какие данные (атрибуты) и какое поведение (методы) будут у его экземпляров.
- **Экземпляр (объект)** — конкретный представитель класса, созданный на его основе. Экземпляры имеют одинаковую структуру, но могут содержать разные значения атрибутов.
- **Атрибут** — любое имя, связанное с объектом (сюда включаются поля, методы, свойства, переменные - про них далее).
- **Поле (атрибут-данные)** — переменная внутри класса или объекта, которая хранит состояние. Поля бывают:  
  - **экземплярные переменные** — принадлежат конкретному объекту и задаются в конструкторе или методах,  
  - **переменные класса** — общие для всех экземпляров.
- **Метод (атрибут-функция)** — функция, определённая внутри класса, описывающая поведение объекта. Методы могут читать и изменять его атрибуты.
- **self** — ссылка на текущий экземпляр, которая позволяет методам обращаться к своим атрибутам и другим методам.
- **Конструктор (`__init__`)** — специальный метод, автоматически вызываемый при создании объекта. Используется для инициализации атрибутов.


*Таким образом, **классы** задают структуру и поведение, а **объекты** являются их конкретными реализациями. **Атрибуты** отвечают за хранение данных, **методы** — за действия, а **`self`** обеспечивает связь между ними внутри каждого экземпляра.*

### Конструктор
Упростить создание экземпляра класса помогает специальный метод `__init__` (но он не защищает от произвольного создания новых полей):

```python
class User:

    def __init__(self, name, rating=0): # как и в обычной функции можно задавать параметры по умолчанию
        self.name = name
        self.rating = rating

    def info(self):
        print(f'Пользователь {self.name} c рейтингом {self.rating}.')

    def info_str(self):
        return f'Пользователь {self.name} c рейтингом {self.rating}.'
```

```python
somebody = User('Вася', rating=50)
somebody.lalala = 6
somebody.info()
```

Защитить экземпляры класса от добавления новых полей и нарушения структуры поможет `__slots__`:

```python
class User:
    __slots__ = ("__name", "__rating")  # разрешаем только эти два атрибута

    def __init__(self, name, rating=0):  # можно задавать значение по умолчанию
        self.__name = name
        self.__rating = rating

    def info(self):
        print(f'Пользователь {self.__name} c рейтингом {self.__rating}.')
```

```python
somebody = User('Вася', rating=50)
# somebody.lalala = 6
somebody.info()
```

```python
somebody.__name = 'Вася'
```

```python
class Contact:

    def __init__(self, name, phone_number):
        self.name = name
        self.phone_number = phone_number

    def info(self):
        return f'{self.name}({self.phone_number})'
```

```python
class Message:

    def __init__(
        self,
        text: str,
        sender: Contact,
        time=None
    ):
        self.text = text
        self.sender = sender
        self.time = time or datetime.now()

    def info(self):
        return f'{self.text}, отправитель: {self.sender.name}, время: {self.time}'
```

## 🟢 2. Инкапсуляция

Инкапсуляция - это принцип ООП, который объединяет **данные (атрибуты)** и **методы** в один объект и контролирует доступ к ним [wiki](https://en.wikipedia.org/wiki/Encapsulation_(computer_programming))

С помощью инкапсуляции можно **скрывать внутренние детали реализации**, предоставляя только нужный интерфейс для работы с объектом. Инкапсуляцию в Python обеспечивают:

* **Защищённые атрибуты** (`_attr`) — предупреждают, что поле не предназначено для прямого внешнего изменения.
* **Приватные атрибуты** (`__attr`) — усложняют доступ извне через *name mangling*.
* **Геттеры и сеттеры** (`get_…`, `set_…`) — контролируют чтение и изменение защищённых полей.
* **Методы класса** — предоставляют управляемый интерфейс для работы с данными.
* **`__slots__`** — ограничивает создание новых атрибутов, защищая внутреннее состояние объекта.

```python
class My:
    def __init__(self):
        self.public_property = 1       # публичное поле, доступно напрямую извне
        self._some_private = 1234      # защищённое поле (convention), можно изменить, но не рекомендуется
        self.__my_private = 12         # приватное поле, доступ извне ограничен через name mangling

    # геттер и сеттер для приватного поля
    def get_my_private(self):
        return self.__my_private

    def set_my_private(self, num):
        self.__my_private = num

    # приватный метод
    def __pri_meth(self):
        print('Hmm..')


a = My()
a.set_my_private(10)                  # изменение приватного поля через сеттер
a.get_my_private()                    # получение приватного поля через геттер
```

## 🟢 3. Полиморфизм

**Полиморфизм** — это способность объектов разных классов реагировать на одно и то же сообщение (вызов метода) по-своему. Иными словами, один интерфейс (название метода) может работать с разными типами объектов, выполняя для каждого подходящую операцию.  

**Полиморфизм нужен для:**

* *Упрощения кода:* можно писать универсальные функции, которые работают с разными объектами.
* *Повышения гибкости:* добавление новых типов объектов не требует изменения существующего кода, если они поддерживают нужный интерфейс.
* *Поддержки принципа открытости/закрытости (Open/Closed Principle) в ООП:* расширять систему можно без модификации старого кода.

Наглядный пример полиморфизма: функция `animal_sound` использует один и тот же вызов `speak()`, но для каждого объекта поведение разное.

```python
class Dog:
    def speak(self):
        return "Гав!"

class Cat:
    def speak(self):
        return "Мяу!"

def animal_sound(animal: Cat | Dog): # это просто функция, не метод какого-то класса
    print(animal.speak())

dog = Dog()
cat = Cat()

animal_sound(dog)  # Выведет: Гав!
animal_sound(cat)  # Выведет: Мяу!
```

Ещё один классический пример: каждый объект shape понимает метод `area()` по-своему, и мы можем обрабатывать разные фигуры одинаковым образом, не проверяя их типы:

```python
class Shape:
    def area(self):
        raise NotImplementedError
```

```python
class TMP(Shape):
    def __init__(self, tmp):
        self.tmp = tmp
    def tmp_method(self, x):
        print(x)
```

```python
example = TMP('привет')
dir(example)
```

```python
example.area()
```

```python
class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2

shapes = [Square(4), Circle(3)]
for shape in shapes:
    print(shape.area())
```

## 🟢 4. Наследование

Наследование позволяет создавать новый класс на основе существующего, повторно используя его код и расширяя функциональность.  

* Новый класс называется **потомком** (или подклассом), а исходный — **родителем** (или суперклассом).  
* Потомок может **переопределять методы родителя**, добавлять новые атрибуты и методы.  
* В Python возможно **множественное наследование**, когда класс наследует функциональность сразу нескольких родителей.  

**Пример.** Классы человек и студент.

```python
class People:
    name = ''
    def set_name(self, name):
        self.name = name

class Student(People): # потомок класса People
    def set_course(self, course):
        self.course = course

    def set_name(self, name):
        super().set_name(name)
        print('!')

some_body = Student()
# dir(some_body)
```

```python
ron = People()
ron.set_name('Рон')
print('----')
gera = Student()
gera.set_name('Гермиона')
gera.set_course(1)
```

```python
for person in [ron, gera]:
    print(person.name)
```

**Пример**. Реализуем различные классы работников, для хранения информации и подсчета заработной платы.

```python
class Worker:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Junior(Worker): # наследуем от класса Worker
    def __init__(self, name, age):
        super().__init__(name, age)
        self.salary = 1000


class Middle(Worker):
    def __init__(self, name, age):
        super().__init__(name, age)
        self.salary = 1000
        self.allowance = 500 # расширяем функционал
```

Использование:

```python
department = [
    Junior('Anji', 25),
    Junior('Pol', 27),
    Junior('Ruby', 34),
    Middle('Sonya', 30),
]
```

```python
total = 0
for worker in department:
    total += worker.salary

    if type(worker) == Middle:
        total += worker.allowance

print(total)
```
